<h4># 1 - Print Function</h4><hr>
<li>É possivel passar argumentos adicionais a função print para modificar como ele funciona</li>


```python
# sep é um argumento que adiciona um separador entre cada string passada dentro da função print
nome = 'Rafael'
idade = 32
print("Meu nome é ", nome, " e eu tenho ", idade, " anos.",sep="'")
# end é um argumento que permite alterar o que é imprimido no final da linha.
print("Meu nome é ", nome, end=" ")
print("Eu tenho ", idade, " anos.")
```

<h4>#3 - Range Function</h4><hr>
<li>A função range gera uma lista de numero começando de um dado valo x e indo até um dado valor n. </li>
<li>A função range retorna um objeto iterator.</li>

```python
# Quando se passa apenas um valor para a função range esse valor determina o ponto de parada, iniciando por padrão em 0.
print(list(range(10)))
# Passando dois argumentos o primeiro argumento é lido como valor inicial e o segundo como valor final.
print(list(range(10, 20)))
# Já com três argumentos o primeiro é o valor inicial, o segundo valor final, e o terceiro o passo.
print(list(range(10, 20, 3)))
# Também é possivel gerar lista decrescente usando passo com valores negativos
print(list(range(50, 0, -5)))
```

<h4>#4 - Map Function</h4><hr>
<li>A função map permite aplicar uma dada função a cada item de um objeto iterável. </li>
<li>Um objeto iterável é qualquer objeto que voce pode percorre-lo por meio de um loop.(arrays, strings, lists, sets..)</li>

```python
# Supnhamos que queremos saber o tamanho de todos os itens da lista de String abaixo
strings = ["Rafael","Ozzy", "Felipe", "Tiago", "Viktor", "Ana", "Oto"]
# Poderiamos percorrer toda a lista de strings e aplicar manualmente a função len() a cada item do iteravel,entretando usando o map conseguimos resolver o problema de forma mais elegante.
# A função map recebe como parameto uma função, que será aplicada a todos os elementos, e o objeto iteravel
lengths = map(len, strings)
print(list(lengths))
# Também é possivel aplicar uma função lambda ao objeto iteravel
lambda_list = map(lambda x: x * x, [1, 2, 3, 4, 5])
print(list(lambda_list))
```

<h4>#5 - Filter Function</h4><hr>
<li>A função filter funciona semelhante a função map,. ela pega cada item de um objeto iteravel passa para um função e caso a função retorne true o filter mantem aquele item no iteravel, mas caso retorne false o item é removido do iteravel.</li>


```python

# Podemos criar uma função que verifica que um item é menor que 5
def shorter_than_5(string):
    return len(string) < 5

strings_filtered = list(filter(shorter_than_5, strings))
print(strings_filtered)
# Novamente tambem podemos usar funções lambda como parametro para o filter
strings2 = ["Rafael","Ozzy", "Felipe", "Tiago", "Viktor", "Ana", "Oto"]
filtered = filter(lambda x: len(x)  > 4, strings2)
print(list(filtered))
```

<h4>#6 - Sum Function</h4><hr>
<li>A função sum retorna a soma de todos os numeros contidos em um objeto iteravel que contenha valores numericos.</li>
<li>Se o objeto iteravel passado não contiver valores numericos a função irá gerar um erro</li>


```python
numeros = {1, 2, 3, 4, 5}
print(sum(numeros))
# É possivel passar um valor inicial a função sum
print(sum(numeros, start=10))
# O valor inicial também pode ser negativo
print(sum(numeros, start=-10))
```

<h4>#7 - Sorted Function</h4><hr>
<li>A função sorted vai ordenar os elementos de um objeto iteravel em ordem crescente ou decrescente dependendo da sua demanda.</li>


```python
numeros = {9, 3, -2, 7, 5, 11, 17, -1, 23, -77}
# Por padrão a função sorted ordena em ordem crescente
ordenado = sorted(numeros)
print(ordenado)
# Entretando é possivel ordenar em ordem decrescente usando o argumento reverse que quando True ordena em ordem decrescente.
ordenado = sorted(numeros, reverse=True)
print(ordenado)
# Outro argumento interessante que pode ser usado é o argumento key, com ele podemos definir qual atributo será usado para ordenação em objetos mais complexos
pessoas = [
    {"nome": "Rafael", "idade": 32},
    {"nome": "Felipe", "idade": 18},
    {"nome": "Lucas", "idade": 21},
    {"nome": "Thyago", "idade": 42},
    {"nome": "André", "idade": 33},
]
# Ordenando pessoas usando como critério de ordenação a idade.
pessoas_ordenadas = sorted(pessoas, key=lambda pessoa: pessoa["idade"])
print(pessoas_ordenadas)
```

<h4>#8 - Enumerate Function</h4><hr>
<li>A função Enumerate, é uma função que resolve o problema de quando percorremos uma lista e precisamos usar ao mesmo tempo o índice e o valor daquela lista, para ficar mais claro vamos demonstrar o uso sem o enumerate, depois usar com ele.</li>
<li>O primeiro valor será o indice e o segundo valor será o valor.</li>


```python
# Vamos imprimir a lista de tarefas mostrando sua o indice + 1 e o conteudo daquela posição
tasks = ["Estudar", "Academia", "Projeto de TCC", "Canal do Youtube", "Pesquisas de artigo"]
print("Solução com Loop:")
for index in range(len(tasks)):
    task = tasks[index]
    print(f"{index + 1}. {task}")

# Agora usando a função enumerate podemos apresentar a mesma solução de forma mais elegante e simples
print("\nSolução com Enumerate: ")
for index, tak in enumerate(tasks):
    print(f"{index + 1}. {tak}")
```

<h4>#9 - Zip Function</h4><hr>
<li>A função Zip combina diferentes objetos iteraveis e automaticamente gerencia quando um objeto iteravel tem mais objetos que o outro.</li>



```python
#Temos duas listas com valores correspondentes, por exemplo, lista de nomes e lista de idades, aonde as posições do nome na lista de nomes tem a idade na posição da lista de idade.
names = ["Rafael", "Ozzy", "Felipe", "Tiago", "Ana", "Oto"]
ages = [32, 78, 23, 31, 17, 56]

print("Solução com Loop:")
for index in range(min(len(names), len(ages))):
    name = names[index]
    age = ages[index]
    print(f"{name} is {age} years old.")

 # Solução usando a função zip para combinar as duas listas
combined_list = (list(zip(names, ages)))
print("\nSolução com zip:")
for name, age in combined_list:
    print(f"{name} is {age} years old.")
```

<h4>#10 - Open Function</h4><hr>
<li>A função open é usada para abrir, arquivos para manipula-los. O primeiro argumento passado é o caminho do arquivo, e o segundo é o modo de abertura.</li>

```python
# Abertura e fechamento manual do arquivo
file = open("files/arquivo.txt", "w+")
file.write("Hello World.\nMy name is Viktor.")
file.close()

# Uma melhor alternativa para a abertura de arquivos é a clausula with, que dispensa o fechamento manual do mesmo.
with open("files/arquivo.txt", "r") as file:
    text = file.read()
    print(text)
```
