1. Construa um algoritmo que, tendo como dados de entrada dois pontos quaisquer no plano, P(x1,y1) e P(x2,y2), escreva a distância entre eles. A fórmula que efetua tal cálculo é:
<br><br><center><img src="formula_01.png"></center>


```python
class Ponto:
    def __init__(self, x, y):
        self.x = x
        self.y = y
```

```python
from math import sqrt
x = int(input('Digite o X do ponto 1: '))
y = int(input('Digite o Y do ponto 1: '))
p1: Ponto = Ponto(x, y)
x = int(input('Digite o X do ponto 2: '))
y = int(input('Digite o Y do ponto 2: '))
p2: Ponto = Ponto(x, y)
distancia = sqrt(pow(p1.x - p2.x, 2)+pow(p1.y - p2.y, 2))
print(f'A distância entre os pontos é: {distancia}.2f')
```
